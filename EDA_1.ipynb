```python
df["City"] = df["City"].fillna("Unknown")
df["Customer_Name"] = df["Customer_Name"].fillna("Unknown")
df["Rating"] = df["Rating"].fillna(df["Rating"].mean())
df["Payment_Mode"] = df["Payment_Mode"].fillna("Unknown")
```

```python
print("Duplicates:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("New Shape:", df.shape)
```

```python
df["Gender"] = df["Gender"].str.strip().str.title()
df["City"] = df["City"].str.strip().str.title()
df["Payment_Mode"] = df["Payment_Mode"].str.strip().str.title()
```

```python
print(df["Gender"].unique())
print(df["City"].unique())
print(df["Payment_Mode"].unique())
```

```python
df["Total Price"] = df["Quantity"] * df["Unit_Price"]

df["Discount Amount"] = df["Total Price"] * df["Discount"]

df["Final Amount"] = df["Total Price"] - df["Discount Amount"]

print(df[["Order_ID", "Product", "Quantity",
          "Total Price", "Discount Amount", "Final Amount"]])
```

```python
print("Total Quantity:", df["Quantity"].sum())

print("Average Unit Price:", df["Unit_Price"].mean())

print("Maximum Unit Price:", df["Unit_Price"].max())

print("Minimum Unit Price:", df["Unit_Price"].min())

print("Average Rating:", df["Rating"].mean())
```

```python
columns = ["Order_ID", "Product", "Quantity", "Rating",
           "Final Amount", "City"]
```
